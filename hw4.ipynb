 Завдання 2 та 3: TP, FP, FN, TN, Precision, Recall та F1-score

```python
# ЗАВДАННЯ 2 та 3: TP, FP, FN, TN, Precision, Recall, F1-score для кожного класу. for class\_id in range(3): tp = 0 fp = 0 fn = 0 tn = 0 for i in range(len(test\_labels)): actual = test\_labels[i] pred = predictions[i] if actual == class\_id and pred == class\_id: tp += 1 elif actual != class\_id and pred == class\_id: fp += 1 elif actual == class\_id and pred != class\_id: fn += 1 else: tn += 1 # Обчислюємо метрики із захистом від ділення на 0 precision = tp / (tp + fp) if (tp + fp) &gt; 0 else 0 recall = tp / (tp + fn) if (tp + fn) &gt; 0 else 0 f1 = ( 2 \* precision \* recall / (precision + recall) if (precision + recall) &gt; 0 else 0 ) print( f"--- Клас {class\_id} ({iris.target\_names[class\_id].upper()}) ---" ) print(f"TP: {tp}, FP: {fp}, FN: {fn}, TN: {tn}") print(f"Precision: {precision:.4f}") print(f"Recall: {recall:.4f}") print(f"F1-score: {f1:.4f}\\n")
```

Відповіді на короткі питання 

1. Чим Precision відрізняється від Recall?Precision (Точність) показує частку справді правильних об'єктів серед усіх, які модель прогнозувала як цей клас ($\frac{TP}{TP+FP}$). Відповідає на питання: "Якщо модель каже, що це клас X, наскільки можна їй вірити?".Recall (Повнота) показує частку знайдених об'єктів серед усіх, які насправді належать до цього класу ($\frac{TP}{TP+FN}$). Відповідає на питання: "Яку частку з усіх реальних квітках класу X модель змогла знайти?".2. Чи може бути висока Accuracy, але поганий Recall для одного класу? Чому?Так, може. Це відбувається при дисбалансі класів (коли одного класу в даних дуже багато, а іншого — мало). Наприклад, якщо в даних 95% клас А і 5% рідкісний клас Б, модель може просто завжди видавати клас А. Загальна Accuracy буде високою (95%), але Recall для рідкісного класу Б дорівнюватиме 0%, оскільки модель не знайде жодного об'єкта цього класу ($FN = \text{усі об'єкти B}$).3.
