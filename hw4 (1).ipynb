ЛЮБІМІ МЕТРИКИ

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# 1. Завантажуємо датасет Iris та ділимо вибірку
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)

# 2. Навчаємо модель KNN
knn = KNeighborsClassifier()
knn.fit(X_train, y_train)

# 3. Робимо прогнози (тут створюються змінні, яких тобі не вистачало!)
predictions = knn.predict(X_test)
test_labels = y_test

# --- ЗАВДАННЯ 1: Accuracy ---
correct = 0
for i in range(len(test_labels)):
    if predictions[i] == test_labels[i]:
        correct += 1

accuracy = correct / len(test_labels)
print("Кількість правильних прогнозів:", correct, "з", len(test_labels))
print(f"Accuracy: {accuracy}\n")

# --- ЗАВДАННЯ 2 та 3: Метрики для кожного класу ---
for class_id in range(3):
    tp = 0
    fp = 0
    fn = 0
    tn = 0

    for i in range(len(test_labels)):
        actual = test_labels[i]
        pred = predictions[i]

        if actual == class_id and pred == class_id:
            tp += 1
        elif actual != class_id and pred == class_id:
            fp += 1
        elif actual == class_id and pred != class_id:
            fn += 1
        else:
            tn += 1

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0

    print(f"--- Клас {class_id} ({iris.target_names[class_id].upper()}) ---")
    print(f"TP: {tp}, FP: {fp}, FN: {fn}, TN: {tn}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1-score:  {f1:.4f}\n")
```

Короткі питання:

### Відповіді на короткі питання:

1. **Precision vs Recall:**
   - **Precision** показує, яка частка серед усіх названих моделлю об'єктів класу є справді правильною (наскільки можна вірити прогнозу).
   - **Recall** показує, яку частку з усіх реальних об'єктів цього класу модель змогла знайти.

2. **Висока Accuracy при поганому Recall:**
   - Так, це можливо при дисбалансі класів. Якщо 95% даних належать до одного класу, модель може завжди передбачати лише його (Accuracy буде 95%), але для рідкісного класу Recall буде 0%.

3. **Результат KNN за F1-score:**
   - Для всіх трьох класів (setosa, versicolor, virginica) F1-score дорівнює 1.0 (100%), оскільки модель на тестовому наборі не зробила жодної помилки.
